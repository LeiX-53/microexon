```python
import pandas as pd
from modlamp.analysis import GlobalAnalysis
from modlamp.descriptors import GlobalDescriptor
import numpy as np
```

```python
df = pd.read_csv('../outputs/miniexon_aa_seq_marked_curated_IDR_pDP.csv')
```

```python
micro = []
rest = []
events = []
whole = []
for event, sequence in zip(df['EVENT'], df['aa_seq_marked']):
    events.append(event)
    micro_seq = sequence.split('#')[1]
    rest_seq = sequence.split('#')[0] + sequence.split('#')[-1]
    whole_seq = sequence.replace('#', '')
    
    desc = GlobalDescriptor(micro_seq)
    desc.calculate_all()
    micro.append(desc.descriptor)
    
    desc = GlobalDescriptor(rest_seq)
    desc.calculate_all()
    rest.append(desc.descriptor)
    
    desc = GlobalDescriptor(whole_seq)
    desc.calculate_all()
    whole.append(desc.descriptor)
```

```python
micro_s = np.squeeze(micro, axis=1)

micro_aa = pd.DataFrame(micro_s, columns=desc.featurenames, index= events)
```

```python
rest_s = np.squeeze(rest, axis=1)

rest_aa = pd.DataFrame(rest_s, columns=desc.featurenames, index= events)
```

```python
whole_s = np.squeeze(whole, axis=1)

whole_aa = pd.DataFrame(whole_s, columns=desc.featurenames, index= events)
```

```python
df.set_index('EVENT').join(micro_aa.join(rest_aa,lsuffix='_miniexon', rsuffix='_rest')).to_csv('../outputs/miniexon_aa_seq_marked_curated_IDR_pDP_aafeature.csv')
```
