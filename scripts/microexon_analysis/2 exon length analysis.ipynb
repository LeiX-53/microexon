```python

import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.ticker as ticker
import numpy as np
from matplotlib.patches import Patch

# ── Segment classification ────────────────────────────────────────────────────
bins   = [0, 21, 51, 90, np.inf]
labels = ['3-21', '24-51', '54-90', '>90']
df_select['Segment'] = pd.cut(df_select['LENGTH'], bins=bins, labels=labels)

seg_counts = df_select['Segment'].value_counts().reindex(labels, fill_value=0)


# ── Colors ───────────────────────────────────────────────────────────────────
palette = {
    '3-21':  '#4C72B0',
    '24-51': '#55A868',
    '54-90': '#DD8452',
    '>90':   '#C44E52',
}
colors = [palette[s] for s in labels]

# ── Plot ──────────────────────────────────────────────────────────────────────
fig, axes = plt.subplots(1, 2, figsize=(13, 5))
fig.suptitle('Exon Length Distribution by Segment', fontsize=14, fontweight='bold')

# --- Left: Bar chart – count per segment ---
ax1 = axes[0]
x_pos = range(len(labels))
bars = ax1.bar(x_pos, seg_counts.values, color=colors, edgecolor='white', width=0.6)
for bar, val in zip(bars, seg_counts.values):
    ax1.text(
        bar.get_x() + bar.get_width() / 2.0,
        bar.get_height() + 0.1,
        str(int(val)),
        ha='center', va='bottom', fontsize=11, fontweight='bold'
    )
ax1.set_xticks(list(x_pos))
ax1.set_xticklabels(labels)
ax1.set_xlabel('Exon Length Segment (bp)', fontsize=11)
ax1.set_ylabel('Number of Exons', fontsize=11)
ax1.set_title('Exon Count per Segment', fontsize=12)
ax1.set_ylim(0, seg_counts.max() + 1.5)
ax1.yaxis.set_major_locator(ticker.MaxNLocator(integer=True))
ax1.spines['top'].set_visible(False)
ax1.spines['right'].set_visible(False)

# --- Right: Pie chart – proportion per segment ---
ax2 = axes[1]
non_zero = seg_counts[seg_counts > 0]
wedge_colors = [palette[s] for s in non_zero.index]
wedges, texts, autotexts = ax2.pie(
    non_zero.values,
    labels=list(non_zero.index),
    colors=wedge_colors,
    autopct='%1.1f%%',
    startangle=140,
    wedgeprops=dict(edgecolor='white', linewidth=1.5),
    textprops=dict(fontsize=11),
)
for at in autotexts:
    at.set_fontsize(10)
    at.set_fontweight('bold')
ax2.set_title('Segment Proportion', fontsize=12)

# --- Shared legend ---
legend_handles = [Patch(facecolor=palette[s], edgecolor='white', label=s) for s in labels]
fig.legend(
    handles=legend_handles,
    title='Segment (bp)',
    loc='lower center',
    ncol=4,
    fontsize=10,
    frameon=False,
    bbox_to_anchor=(0.5, -0.04)
)

plt.tight_layout(rect=[0, 0.05, 1, 1])
# plt.savefig('exon_length_segments.png', dpi=150, bbox_inches='tight')
# plt.show()
# print('Plot saved to exon_length_segments.png')
```
