```python
import pandas as pd
```

```python
df = pd.read_csv('../outputs/miniexon_aa_seq_curated.csv')
```

```python
df['aa_seq_marked'] = df.apply(lambda x:('#'+x['exon_aa_seq'].split('*')[0]+'#').join(x['aa_seq'].split(x['exon_aa_seq'].split('*')[0])), axis=1)




```

```python
df_cur = pd.read_csv('../source_data/metadata.csv')
```

```python
cur_map = df_cur.drop_duplicates("EVENT").set_index("EVENT")["aa_seq_marked"]
df_map = df.drop_duplicates("EVENT").set_index("EVENT")["aa_seq_marked"]

result = [
    cur_map.get(event, df_map.get(event))
    for event in df["EVENT"]
]
```

```python
df['aa_seq_marked'] = result
```

```python
df['#'] = df.apply(lambda x:x['aa_seq_marked'].count('#'), axis=1)
```

```python
df[df['#'] !=2]
```

```python
df.to_csv('../outputs/miniexon_aa_seq_marked_curated.csv')
```
